```python
# 함수
def plusTwo(num):
  print(num + 2)
  return 'End Plus'

plusTwo(2)

def sayHello(name, num):
  print(f'안녕하세요. {name}님\n' * num)
  return 'End'

sayHello('김민아', 2)
```

```python
def makeMsg(list):
  messages = []
  for i, item in enumerate(list):
    messages.append(item)
  return messages

list = [
    {'role':'user', 'content':'비밀번호를 잊어버렸어요.'},
    {'role':'assistant', 'content':'답변을 할게요.'}
]
messages = makeMsg(list)
print(messages)
print(len(messages))
```

```python
# if - 질문인가요?
def is_question(str):
  return str[-1] == '?' or '요' in str[-3:]

prompt = '이게 질문이라고 하나요오오'
is_question(prompt)
```

```python
# try except
# print(10 / 0)
try:
  result = 10 / 0
except ZeroDivisionError:
  print('0으로 나눌 수 없습니다.')
  result = None

print(result)
```
