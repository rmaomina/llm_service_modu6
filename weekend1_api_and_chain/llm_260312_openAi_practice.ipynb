```python
# temperature (얼마나 다양한 단어를 선택할까?)
# 낮은 값에서는 똑같은 값, 높은 값에서는 다른 값.. -- 창의성?!
question = '인공지능의 미래에 대해 한 문장으로 말해봐!'
for i in range(3):
  response = client.chat.completions.create(
      model = 'gpt-4o-mini',
      messages = [{'role': 'user', 'content': question}],
      temperature = 0
  )
  print(f'{i+1}: {response.choices[0].message.content})')
```

```python
# temperature
question = '인공지능의 미래에 대해 한 문장으로 말해봐!'
for i in range(3):
  response = client.chat.completions.create(
      model = 'gpt-4o-mini',
      messages = [{'role': 'user', 'content': question}],
      temperature = 1.5
  )
  print(f'{i+1}: {response.choices[0].message.content})')
```

```python
# reasoning이 접목된 5모델, 그래서 5모델에서는 temperature 설정이 없다. 1로 고정.
# temp를 높이면, reasoning 로직에 영향이...
question = '고양이의 특성에 대해 한 문장으로 알려줘'
for temp in [0, 0.5, 1.0]:
  for i in range(2):
    response = client.chat.completions.create(
      model = 'gpt-4o-mini',
      messages = [{'role': 'user', 'content': question}],
      temperature = temp
    )
    print(f'{i+1}: {response.choices[0].message.content})')
```

```python
# max_tokens
response = client.chat.completions.create(
  model = 'gpt-4o-mini',
  messages = [{'role': 'user', 'content': '대한민국에 대해 알려줘'}],
  max_tokens = 50
)
response.choices[0].message.content
```

```python
# temperature, max_tokens 3개씩 조합해서 연습해 보세요.
temp = [0, 0.5, 1.0]
max_tokens = [50, 100, 200]
question = '독일의 초등학교 교육 시스템에 대해서 설명해 주세요.'

for i in range(len(temp)):
  for j in range(len(max_tokens)):
    response = client.chat.completions.create(
      model = 'gpt-4o-mini',
      messages = [{'role': 'user', 'content': question}],
      max_tokens = max_tokens[j],
      temperature = temp[i]
    )
    print(f'Temperature: {temp[i]}, Max Tokens: {max_tokens[j]}\n{response.choices[0].message.content}\n--------------------')
```

```python
# top_p (얼마나 보여줄까? 누적 확률)
# temperature과 top_p는 보통 같이 사용하지 않는다.
```

```python
# Penalty
# frequency_penalty: 반복에 대한 패널티, 같은 단어가 계속 나오지 않도록. 0-1,
# 0은 반복 신경쓰지 않음, 1은 반복하면 안됨(반복 방지)
# presence_panalty: 기존에 나왔던 단어 사용하지 않음, 1: 적극적으로 새로운 주제나 단어..
question = '겨울을 주제로 짧은 시를 써 줘.'
response = client.chat.completions.create(
  model = 'gpt-4o-mini',
  messages = [{'role': 'user', 'content': question}],
  top_p = 0.1 # 상위 10%의 단어만 가지고 생성 요청
)
response.choices[0].message.content
```

```python
# stop 사용법
question = '프로그래밍 언어 10개를 나열해 줘'
response = client.chat.completions.create(
  model = 'gpt-4o-mini',
  messages = [{'role': 'user', 'content': question}],
  stop = ['6. ']
)
print(response.choices[0].message.content)
print(response) #finish_reason='stop'
print('---------')
question = '프로그래밍 언어 10개를 나열해 줘'
response = client.chat.completions.create(
  model = 'gpt-4o-mini',
  messages = [{'role': 'user', 'content': question}]
)
print(response.choices[0].message.content)
print(response) #finish_reason='stop'
```

```python
# 세 턴 이상의 대화를 만들어 보세요.
messages = [
    {'role': 'system', 'content': '당신은 의사입니다. 환자를 진단하고 있어요. 친절하고 간결하게 설명해 주세요'},
]

def send_message_and_print(user_content):
  messages.append({'role': 'user', 'content': user_content})
  print(f"**환자:** {user_content}")

  response = client.chat.completions.create(
    model = 'gpt-4o-mini',
    messages = messages,
  )

  assistant_content = response.choices[0].message.content
  messages.append({'role': 'assistant', 'content': assistant_content})
  print(f"**의사:** {assistant_content}\n")

# 대화 시작
send_message_and_print('머리가 아파요')
send_message_and_print('수면 부족인 것 같은데, 그럼 배도 아픈가요? 어떻게 하면 좋아질까요?')
send_message_and_print('자기전에 마사지도 도움이 되나요?')
```

```python
# 스트리밍 형식
messages = [
    {'role': 'user', 'content': '파이썬의 장점 세 가지를 얘기해 줘 '},
]
response = client.chat.completions.create(
  model = 'gpt-4o-mini',
  messages = messages,
  stream = True
)
full_answer = ''
for chunk in response:
  text = chunk.choices[0].delta.content
  if text:
    print(text, flush = True, end = '')
    full_answer += text
full_answer
```
