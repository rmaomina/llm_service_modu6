````python
# ════════════════════════════════════════════════════════════════
# 📦 전 주차 통합 패키지 설치
# -q: quiet 모드 (설치 로그 최소화)
# 이미 설치된 패키지는 건너뛰므로, 매번 실행해도 안전합니다
# ════════════════════════════════════════════════════════════════

# ── [1~3주차] LangChain 기초 + RAG + 프롬프트 + 메모리 ──────────
#   openai            : OpenAI API 클라이언트
#   langchain         : LangChain 코어 프레임워크
#   langchain-openai  : LangChain의 OpenAI 통합 (ChatOpenAI, OpenAIEmbeddings)
#   langchain-community: FAISS 등 커뮤니티 통합 모듈
#   faiss-cpu         : Facebook AI 벡터 유사도 검색 (CPU 버전)
#   python-dotenv     : .env 파일에서 환경변수 로드
#   gradio            : 웹 UI를 빠르게 만드는 라이브러리
#   numpy             : 수학 연산 (코사인 유사도 계산 등)
!pip install -q openai langchain langchain-openai langchain-community faiss-cpu python-dotenv gradio numpy

# ── [4~6주차] RAG 검색 성능 평가 + 하이브리드 검색 + 추천 시스템 ──
#   rank-bm25         : BM25 키워드 검색 (하이브리드 검색에 사용)
#   langchain-chroma  : Chroma 벡터 DB (FAISS 대안)
#   chromadb          : Chroma 벡터 데이터베이스
#   sentence-transformers : HuggingFace 임베딩 모델 (다국어 지원)
#   datasets          : HuggingFace 데이터셋 로더
#   pandas            : 데이터프레임 처리
#   matplotlib        : 그래프/시각화
!pip install -q rank-bm25 langchain-chroma chromadb sentence-transformers datasets pandas matplotlib

# ── [7~9주차] Tool Calling + LangGraph 에이전트 ──────────────────
#   langgraph         : LangGraph - 상태 기반 에이전트 프레임워크
#   langchain-anthropic: Claude 모델 연동
#   langchain-google-genai: Gemini 모델 연동
#   tavily-python     : Tavily 웹 검색 API (에이전트용 검색 도구)
#   langchainhub      : LangChain 커뮤니티 프롬프트 허브
!pip install -q langgraph langchain-anthropic langchain-google-genai tavily-python langchainhub

# ── [10~12주차] 비정형 문서 + 멀티모달 + Graph RAG ─────────────
#   unstructured      : 비정형 문서(PDF, DOCX 등) 파싱
#   pdf2image         : PDF → 이미지 변환
#   pymupdf           : PDF 텍스트 추출
#   pillow            : 이미지 처리
#   neo4j             : Neo4j 그래프 DB 드라이버
#   langchain-neo4j   : LangChain Neo4j 통합
!pip install -q unstructured pdf2image pymupdf pillow neo4j langchain-experimental

print("✅ 전체 패키지 설치 완료!")
````

## ⚙️ Step 2: 환경 설정 + 라이브러리 임포트

아래 셀은 **매번 노트북을 열 때마다** 실행해야 합니다.  
(Colab은 세션이 끊기면 메모리가 초기화되기 때문)

````python
# ════════════════════════════════════════════════════════════════
# ⚙️ 환경 설정 (매 세션마다 실행)
# ════════════════════════════════════════════════════════════════
import os
from google.colab import userdata

# ── API 키 설정 ──────────────────────────────────────────────────
# Colab Secrets에서 키를 가져와 환경변수로 등록
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')

# ── [1~3주차] 기본 라이브러리 ────────────────────────────────────
from openai import OpenAI                              # OpenAI 원시 API
from langchain_openai import ChatOpenAI, OpenAIEmbeddings  # LangChain OpenAI 래퍼
from langchain_core.prompts import ChatPromptTemplate      # 프롬프트 템플릿
from langchain_core.output_parsers import StrOutputParser  # LLM 응답 → 문자열 변환
from langchain_core.documents import Document              # 문서 객체
from langchain_core.runnables import (
    RunnablePassthrough,   # 입력을 그대로 통과
    RunnableLambda,        # 일반 함수를 LCEL 체인에 연결
    RunnableParallel,      # 여러 체인을 병렬 실행
)
from langchain_community.vectorstores import FAISS         # 벡터 검색 엔진

import numpy as np
import time

# ── 클라이언트 초기화 ─────────────────────────────────────────────
client = OpenAI()                                          # OpenAI 원시 API 클라이언트
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)       # LangChain LLM (저온도 = 일관된 답변)
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")  # 임베딩 모델 (1536차원)

print("✅ 환경 설정 완료!")
print(f"   LLM: gpt-4o-mini | Embeddings: text-embedding-3-small")
````

---
## ✅ Step 3: 동작 확인

API 키와 라이브러리가 정상 작동하는지 확인합니다.

````python
# ════════════════════════════════════════════════════════════════
# ✅ 동작 확인 테스트
# ════════════════════════════════════════════════════════════════

# 1. OpenAI API 직접 호출 테스트
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "안녕하세요! 한 줄로 인사해주세요."}],
    max_tokens=50
)
print(f"1️⃣ OpenAI API: {response.choices[0].message.content}")

# 2. LangChain LLM 테스트
chain = ChatPromptTemplate.from_messages([
    ("user", "{question}")
]) | llm | StrOutputParser()
print(f"2️⃣ LangChain:  {chain.invoke({'question': '1+1은?'})}")

# 3. 임베딩 테스트
vec = embeddings.embed_query("테스트")
print(f"3️⃣ Embedding:  {len(vec)}차원 벡터 생성 ✓")

# 4. FAISS 테스트
test_docs = [Document(page_content="테스트 문서", metadata={"id": "test"})]
test_vs = FAISS.from_documents(test_docs, embeddings)
print(f"4️⃣ FAISS:      벡터 스토어 생성 ({test_vs.index.ntotal}개) ✓")

print("\n🎉 모든 테스트 통과! 실습을 시작하세요.")
````
